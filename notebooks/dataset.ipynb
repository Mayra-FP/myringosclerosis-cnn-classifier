```python
# Preparacion del entorno de trabajo para descarga de datos y acceso a carpeta
from google.colab import drive # montar drive
import kagglehub # accecer a la API de datos kaggle
import shutil # copiar archivo de API kaggle a Drive
import os # trabajar con directorios
```

```python
# Acceso a drive
drive.mount('/content/drive') # genero el acceso a mi google drive para importar los archivos necesarios
```

```python
# drive path
drive_path = "/content/drive/MyDrive/PFM/Datasets"

# Establecer un directorio donde guardar
os.makedirs(drive_path, exist_ok=True)
os.getcwd() # ver directorio
```

```python
# Download dataset de kaggle
kaggle_path = kagglehub.dataset_download("ucimachinelearning/otoscopic-image-dataset")
```

```python
# Copiar datasets de kaggle a mi drive
shutil.copytree(kaggle_path, drive_path, dirs_exist_ok= True)

print("Dataset copiado a:", drive_path)
```

```python
# Dividir en test, entrenamiento y validación los conjuntos Myringosclerosis y Normal

# Ruta original y destino
orig_path = os.path.join(drive_path, "Otoscopic_Data")
dest_path = os.path.join(drive_path, 'imagenes_otoscopia')

# Porcentajes
train_pct = 0.7
val_pct = 0.15
test_pct = 0.15

# Solo usar estas clases
CLASSES = ['Myringosclerosis', 'Normal']

# Crear carpetas destino si no existen
for split in ['train', 'val', 'test']:
    for class_name in CLASSES:
        os.makedirs(os.path.join(dest_path, split, class_name), exist_ok=True)

# Import necessary libraries
import random
from sklearn.model_selection import train_test_split

# Separar por clase
for class_name in CLASSES:
    class_path = os.path.join(orig_path, class_name)
    if not os.path.isdir(class_path):
        continue

    images = os.listdir(class_path)
    random.shuffle(images)

    train_imgs, temp_imgs = train_test_split(images, train_size=train_pct)
    val_imgs, test_imgs = train_test_split(temp_imgs, test_size=test_pct / (val_pct + test_pct))

    # Copiar imágenes
    for img in train_imgs:
        shutil.copy(os.path.join(class_path, img), os.path.join(dest_path, 'train', class_name, img))

    for img in val_imgs:
        shutil.copy(os.path.join(class_path, img), os.path.join(dest_path, 'val', class_name, img))

    for img in test_imgs:
        shutil.copy(os.path.join(class_path, img), os.path.join(dest_path, 'test', class_name, img))
```
